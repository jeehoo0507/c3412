## GCD

A,B GCD = B,r GCD

A = aG, B = bG (a,b 서로소)

A = qB + r
aG = qbG + r
r = (a-qb)G

B, r에서 반약 r = 0이 되면?
A = qB + 0이라는 뜻
곧 B가 최대공약수
최대공약수는 GCD(A,B) = GCD(B,r) = GCD(B,0) = B


```c
#include <stdio.h>
int gcd(int a, int b){
    while(b!=0){
        int r = a%b;
        a = b;
        b = r; 
    }

    return a;
}

int main(){
    int a,b;
    scanf("%d %d", &a, &b);
    printf("%d", gcd(a, b));

    return 0;
}
```

```c
#include <stdio.h>
int lcm(int a, int b){
    int x = a, y = b;
    while(b!=0){
        int r = a%b;
        a = b;
        b = r;
    }
    return (x*y)/a;
}

int main(){

    int a, b;
    scanf("%d%d", &a, &b); //잡다 %d%d로 붙여 써도 입력값은 공백이나 줄바꿈으로 구분해서 받아짐 만약에 1218 이런식으로 하면 첫 %d에 들어가짐 그 후 기다림
    printf("%d", lcm(a,b));
    return 0;
}
```

```c
#include <stdio.h>
int gcd(int a, int b){
    if(b==0) return a;
    return gcd(b,a%b);
}

int lcm(int a, int b){
    return (a*b)/gcd(a,b);
}

int main(){
    int a, b;
    scanf("%d %d", &a, &b);
    printf("%d", lcm(a,b));

    return 0;
}
```
