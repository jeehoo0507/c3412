난수 범위 설정

random = rand() % (최댓값 - 최솟값 + 1) + 최솟값 <-- 중요>

0~45
random = rand() % 46

30~75
random = rand() % 46 + 30

```c
//로또 생성기

#include <stdio.h>
#include <stdlib.h>
#include <time.h>

int main(){
    int i= 0 ,j = 0;
    int random;
    int randnum[6];
    srand(time(NULL));

    for(i =0;i<6;i++){
        random = rand()%45+1;

        for(j=0;j<i;j++){
            if(randnum[j]== random) {
                i--;
                break; //가장 가까운 반복문 탈출
            }
        }

        if(i==j) randnum[i] = random;
    }

    for(int i=0;i<6;i++){
        printf("%d ",randnum[i]);
    }
    return 0;
}
```

| 단계 | `i` | `random` | `j` 변화 | 최종 `j` | 결과 |
|---|---|---|---|---|---|
| 1 | 0 | 7 | 검사 없음 | 0 | `7` 저장 |
| 2 | 1 | 12 | 0 → 1 | 1 | `12` 저장 |
| 3 | 2 | 7 | 0 → break | 0 | 중복, 재시도 |
| 4 | 2 | 25 | 0 → 1 → 2 | 2 | `25` 저장 |
| 5 | 3 | 31 | 0 → 1 → 2 → 3 | 3 | `31` 저장 |
| 6 | 4 | 12 | 0 → 1 → break | 1 | 중복, 재시도 |
| 7 | 4 | 40 | 0 → 1 → 2 → 3 → 4 | 4 | `40` 저장 |
| 8 | 5 | 3 | 0 → 1 → 2 → 3 → 4 → 5 | 5 | `3` 저장 |
