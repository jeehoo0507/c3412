증감 연산자

후위 vs 전위

*(ptr++) : 값 반환 후 증가 <- 이거 개중요

```c
// swap

#include <stdio.h>
void swap(int *a, int *b){
    int temp = *a;
    *a = *b;
    *b = temp;
}

int main(){
    int a, b;
    scanf("%d %d",&a,&b);

    swap(&a,&b);

    printf("%d %d", a,b);
    return 0;
}
```
