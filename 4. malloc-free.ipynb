## 동적 메모리 할당

stdlib.h
필요한 메모리를 할당

기본 형식
(형변환)malloc(할당할 메모리 크기) <- 할당한 메모리의 주소를 반환

free(할당한 메모리의 주소) <- 할당한 메모리를 해제

"메모리 주소" 이다 보니 아래와 같이 배열 처럼 쓰는 것이 가능하다

int* ptr = (int*)malloc(sizeof(int) * 5); 
prt[1] = 10; // ptr[1]은 *(ptr + 1)과 동일

malloc은 메모리 할당 실패 시 NULL을 반환

이딴 것도 가능함 ptr이 가리키는 자료형의 크기를 sizeof(*ptr)로 구할 수 있음
(sizeof(ptr) 안하게 조심 이건 그냥 포인터 크기임)

int* ptr = (int*)malloc(sizeof(int));
int* ptr = (int*)malloc(sizeof(*ptr));
int* ptr = (int*)malloc(sizeof *ptr); <- 괄호 생략 가능

int *ptr = (int *)malloc(sizeof *ptr * 10);
