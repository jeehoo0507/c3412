## 5. 문자열

문자열 = 문자를 저장하는 배열

항상 '\0'(널문자)으로 끝남 
그래서 문자열의 길이보다 1 더 크게 배열을 만들어야 한다.


큰 따옴표로 초기화 가능하다.
char str[6] = "APPLE"; // str[5] = '\0' 자동으로 들어감

배열 크기 > 문자열 : 남는 요소는 '\0'로 초기화됨

%s

printf("%s", str); <- '\0' 전까지 출력
scanf("%s", str); <- 공백 문자( 띄어쓰기, 엔터, 탭) 등 전까지 입력

scanf는 매개변수가 "포인터" 아므로 주소를 입력 받음
str은 &str[0]과 같아서 이미 주소라 위와 같이 사용 가능

```c
#include <stdio.h>

int main(){
    char str[6] = {0}; // 이거 가능함 \0 쓰고 싶으면 {'\0'} 써야함
    return 0;
}
```

```c
// 값 각각 할당 할 때 <- 마지막에 '\0' 넣어줘야함

#include <stdio.h>

int main(){
    char str[6];

    for(int i = 0; i<5;i++){
        scanf("%c", &str[i]); // %c로 받는 거 중요함
    }

    str[5] = '\0';

    printf("%s", str);

    return 0;
}
```

```c
#include <stdio.h>

int main(){

    char a[3];
    scanf("%c", &a[0]);
    scanf("%c", &a[1]);
    scanf("%c", &a[2]);

    printf("%s",a);

    return 0;
}

// 이런식으로 짜면 줄바꿈까지 먹네;; 시험에는 안나올듯
```

문자열 리터럴

"  " 문자열 리터럴도 뒤에 보이지 않는 '\0'이 들어가 있음
마찬가지로 첫번째 주소를 반환해줌
문자열 리터럴은 "읽기 공간"에 저장되기에

일반적으로 char str[6] = "APPLE"; 로 존재 자체가 복사인 문법에서는 상관없지만
char* ptr = "APPLE"; 이면 읽기 공간을 가리키기 떄문에 값을 읽기만 가능하고, 값을 바꾸려고 하면 오류 발생

가리키는 건 가능하기 때문에 그냥 가리키는 값을 바꾸는건 가능

```c
#include <stdio.h>

int main(){
    char *str = "apple";
    printf("%s\n", str); // %s 는 첫번째 주소부터 '\0'까지임 그래서 첫번째 주소 넣어주면 됨

    str = "banana";

    printf("%s", str);


    return 0;
}
```

```c
// 문자 몇번 등장???

#include <stdio.h>

int main(){

    char str[] = "The quick brown fox jumps over the lazy dog";
    char ch;

    int count = 0;

    printf("문자를 입력하세요 : ");
    scanf("%c", &ch); // scanf("문자를 입력하세요 %c : ",&ch) 이렇게 쓰면 문자를 입력하세요 %c : 를 입력해야함 <- 이거 맞음? 

    for(int i=0;i<sizeof(str)/sizeof(str[0]);i++){
    //for(int i=0; str[i]!='\0'; i++) 이것도 가능
        if(str[i] == ch) count++;
    }

    printf("%d", count);

    return 0;
}
```

```c
// 소문자 대문자 변환
// 아스키 'a' = 97 'A' = 65 
// 32 차이남

#include <stdio.h>

int main(){
    char str[] = "thE qUIck brOWN foX juMPs OVER thE lAzy DOg";

    for(int i=0;i<sizeof(str)/sizeof(str[0]);i++){
        if(str[i]>'a' && str[i]<'z') str[i] -= 32;
        if(str[i]>'A' && str[i]<'Z') str[i] += 'a'-'A'; // 32 = 'a'-'A'
    }

    printf("%s", str);

    return 0;
}
```
