배열2 - 형성평가8 (파스칼 삼각형)

https://jungol.co.kr/problem/12005?cursor=IjU2OCIsOCwyMQ==

```c
#include <stdio.h>

int main(){
    int arr[10][10] = {0};

    int n;
    scanf("%d", &n);

    for(int i = n-1 ;i>=0;i--){
        for(int j = 0; j < n - i ; j++){
            if(j==0 || j == n - i - 1){
                arr[i][j] = 1;
            }
            else{
                arr[i][j] = arr[i+1][j-1] + arr[i+1][j];
            }
        }
    }

    for(int i = 0; i < n ; i++){
        for(int j = 0; j < n ; j++){
            if(arr[i][j] == 0 ) printf(" ");
            else printf("%d ",arr[i][j]);
        }
        printf("\n");
    }

    return 0;
}
```

풀이

우선 입력
for문을 생각함

전체 arr의 크기는 10 x 10이고,
이 문제의 경우에는 특이하게 위에서부터 값을 채우는 것이 아닌 아래서부터 값을 채워야함
간단하게 생각해보면 입력값이 n이라면 n-1번 행에 채우고 그 뒤로는 행을 줄이면서 넣어주면 됨

그래서 바깥 for문은 n-1부터 0까지 감소하게 하고
안쪽 for문은 원래 0부터 n-1까지 증가할때를 생각해야하는데 이걸 뒤집은 거니까

0부터 할꺼면 n - i - 1까지 해보면 됨
그다음 양끝값은 1로 초기화하고 
나머지는 자신 - 1행의 값의 자신의 -1,0 인 지점의 열을 더해주면 됨

출력은 nxn을 하되 채워지지 않은 것들은 0으로 채운 뒤 0이면 " " 공백 출력
